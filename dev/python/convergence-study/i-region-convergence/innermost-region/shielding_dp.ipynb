```python
fs = {"rad1" : lambda data_dict, m0 : (data_dict["a"][1] - data_dict["a"][0]),
      "rr0s" : lambda data_dict, m0 : (data_dict["h"] - data_dict["d"][1])/(data_dict["h"] - data_dict["d"][0]),
      "rr1s" : lambda data_dict, m0 : (data_dict["h"] - data_dict["d"][2])/(data_dict["h"] - data_dict["d"][1]),
      "hrs" : lambda data_dict, m0 : (data_dict["h"])/(data_dict["h"] - data_dict["d"][2])}

def plot_varied_cases(fixed_prob_lsts, idx, hydro, f = None, xlab = "X", error = 0.01):
  subset = fixed_prob_lsts[idx].copy()
  val_func = lambda data: data[data["m0s"][0]][f"convergence point {error:.2g} " + hydro]
  subset.sort(key = val_func)
  vals = [val_func(data) for data in subset]
  filter_and_plot_shapes(subset, lambda data_dict, m0 : True, 
                        lambda data_dict, m0 : get_plasma_color(data_dict[m0][f"convergence point {error:.2g} " + hydro], min_value = min(vals), max_value=max(vals)),
                        ppr = 10, figsize_per_plot = (2, 2))
  if f is not None:
    xs = [f(data_dict, data_dict["m0s"][0]) for data_dict in subset]
    ys = [data_dict[data_dict["m0s"][0]][f"convergence point {error:.2g} " + hydro] for data_dict in subset]
    plt.scatter(xs, ys)
    plt.xlabel(xlab)
    plt.ylabel("convergence point 0.01")
    plt.title(hydro)
    plt.show()
```

```python
def plot_alpha_beta_against_2(data, f1, f2, hydro, f1_name, f2_name):
  subsets = subdivide_lst(data, f1)
  alpha_data, beta_data = [], []
  for subset in subsets:
    x = f1(subset[0], subset[0]["m0s"][0])
    pack = multi_fit_parameters(subset, f2, hydro, local_maxima = False, plot_comparison = False, print_params = False,
                                plot_multi_log_comparison = False, plot_multi_params = False, nmk_max = 150)
    for i in range(len(pack[0])):
      y, alpha, beta = pack[0][i], pack[1][i], pack[2][i]
      alpha_data.append([x, y, alpha])
      beta_data.append([x, y, beta])
  plot_3tuples_2d(alpha_data, cmap='plasma', xlabel = f1_name, ylabel = f2_name, clabel = "alpha", title = "alpha, " + hydro)
  plot_3tuples_2d(beta_data, cmap='plasma', xlabel = f1_name, ylabel = f2_name, clabel = "beta", title = "beta, " + hydro)

def plot_alpha_beta_variant1_helper(data, f1_name, f2_name, ylabel, title, exp_fit = False):
  f = lambda x, a, b, c: a * np.exp(-b * x) + c
  for pack in data:
    plt.plot(pack[0], pack[1], color = pack[2], label = f1_name + f" = {pack[3]:.3g}")
    if exp_fit:
      try:
        popt, pcov = curve_fit(f, pack[0], pack[1], p0=(2, 1, 0.5))
        ys_calc = [f(x, *popt) for x in pack[0]]
        label = f"{pack[3]:.3g}, ({popt[0]:.3g}, {popt[1]:.3g}, {popt[2]:.3g})"
        plt.plot(pack[0], ys_calc, color = pack[2], linestyle = "--", label = label)
      except (RuntimeError, ValueError, OverflowError): pass

  plt.legend(loc='center left', bbox_to_anchor=(1.0, 0.5))
  plt.xlabel(f2_name)
  plt.ylabel(ylabel)
  plt.title(title)
  plt.show()
  
def plot_alpha_beta_variant1(data, f1, f2, hydro, f1_name, f2_name, exp_fit = False):
  subsets = subdivide_lst(data, f1)
  prototypes = [subset[0] for subset in subsets]
  f1_vals = [f1(cf, cf["m0s"][0]) for cf in prototypes]
  colors = [get_plasma_color(val, min_value = min(f1_vals), max_value=max(f1_vals)) for val in f1_vals]
  alpha_data, beta_data = [], []
  for i in range(len(subsets)):
    subset, color = subsets[i], colors[i]
    pack = multi_fit_parameters(subset, f2, hydro, local_maxima = False, plot_comparison = False, print_params = False,
                                plot_multi_log_comparison = False, plot_multi_params = False, nmk_max = 150)
    alpha_data.append([pack[0], pack[1], color, f1_vals[i]])
    beta_data.append([pack[0], pack[2], color, f1_vals[i]])
  plot_alpha_beta_variant1_helper(alpha_data, f1_name, f2_name, "alpha", "alpha, " + hydro, exp_fit)
  plot_alpha_beta_variant1_helper(beta_data, f1_name, f2_name, "beta", "beta, " + hydro, exp_fit)
  
```
