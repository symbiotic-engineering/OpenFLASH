Claim: The metrics (h-d0)/radwidth0, (h-d1)/(h-d0), and max((d[i] - d[0])/(a[i-1]-a[0])) accounts for almost all of the variation in convergence curves for added mass in the innermost region. This is shown through histograms.

Furthermore, data sets are shown here where each value individually varies while keeping the others constant, to try and show the shapes of individual dependencies. Multiple plots should be shown for each variable, to show how shapes stay the same or differ across configurations.

```python
file1 = "data/histogram-4.pkl"
file4 = "data/histogram-5.pkl"
file2 = "data/predetermined-2.pkl"
file3 = "data/predetermined-7.pkl"
file5 = "data/histogram-6.pkl"

def right_ratio(cf):
  return (cf["h"]-cf["d"][1])/(cf["h"]-cf["d"][0])

def variable_values(cf):
  output = {"(h-d0)/radwidth0" : (cf["h"]-cf["d"][0])/(cf["a"][0]),
            "(h-d1)/(h-d0)" : (cf["h"]-cf["d"][1])/(cf["h"]-cf["d"][0]),
            "max((di-d0)/(a(i-1)-a0))" : max([(cf["d"][i] - cf["d"][0])/(cf["a"][i-1] - cf["a"][0]) for i in range(2, len(cf["d"]))])}
  if right_ratio(cf) > 1:
    output["radwidth0/a1"] = cf["a"][0]/cf["a"][1]
  return output
```

```python
# Generate a dicionary of ratios.
def hmdi(data_dict, i):
  return data_dict["h"] - data_dict["d"][i]

def di(data_dict, i):
  return data_dict["d"][i]

def radwidthi(data_dict, i):
  return data_dict["a"][i] - (0 if i == 0 else data_dict["a"][i-1])

def radrange(data_dict, i, j):
  return sum([radwidthi(data_dict, k) for k in range(i+1, j + 1)])

dim1s = {"h" : lambda data_dict, m0: data_dict["h"],
        "(1/m0)" : lambda data_dict, m0: 1/m0}

for i in range(3):
  dim1s["d" + str(i)] = lambda data_dict, m0, i=i : di(data_dict, i)
  dim1s["(h-d" + str(i) + ")"] = lambda data_dict, m0, i=i: hmdi(data_dict, i)
  dim1s["radwidth" + str(i)] = lambda data_dict, m0, i=i: radwidthi(data_dict, i)
  dim1s["a" + str(i)] = lambda data_dict, m0, i=i: radrange(data_dict, -1, i)
  for j in range(i + 2, 3):
    dim1s["(a" + str(j) + "-a" + str(i) + ")"] = lambda data_dict, m0, i=i, j=j : radrange(data_dict, i, j)

dim0s = {} # size = 15**2
for key1 in dim1s.keys():
  for key2 in dim1s.keys():
    if key1 == key2:
      pass
    else:
      dim0s[key1 + "/" + key2] = (lambda data_dict, m0, k1 = key1, k2 = key2 : dim1s[k1](data_dict, m0)/dim1s[k2](data_dict, m0))
```

```python
x_funcs = {"radwidth0/a1" : dim0s["radwidth0/a1"]}
y_funcs = {"0.01 am" : (lambda data_dict, m0 : data_dict[m0]["convergence point 0.01 am"])}
x_func, y_func = x_funcs["radwidth0/a1"], y_funcs["0.01 am"]
for prob_set in hist_prob_sets:
  xs = [x_func(prob, m0) for prob in prob_set for m0 in prob["m0s"]]
  ys = [y_func(prob, m0) for prob in prob_set for m0 in prob["m0s"]]
  if max(ys) != min(ys):
    plt.scatter(xs, ys)
    plt.xlabel("radwidth0/a1")
    plt.ylabel("0.01 am")
    plt.title(variable_values(prob_set[0]))
    plt.show()
```

```python
def plot_grouped_dicts(f1, floats, all_hist_data, hydro, error, function_name):
    """
    f1, f2, f3 : callable
        Functions applied to dicts.
    floats : list of float
        Reference float values.
    list_of_dict_lists : list of list of dict
        Each sublist is a group of dicts.
    """
    region_count = lambda cf : len(cf["d"])
    cp = lambda cf : cf[cf["m0s"][0]][f"convergence point {error:.2g} " + hydro]
    
    # assign each float a unique color
    colors = plt.cm.tab10(np.linspace(0, 1, len(floats)))
    float_to_color = {flt: col for flt, col in zip(floats, colors)}

    plt.figure(figsize=(8, 6))

    for flt in floats:
        # collect dict-lists whose first dict maps (via f1) to this float (within 0.01%)
        associated_lists = [
            prob_set for prob_set in all_hist_data
            if abs(f1(prob_set[0]) - flt) <= 0.0001 * flt]
        
        legend_designated = False
        for prob_set in associated_lists:
            values = [cp(cf) for cf in prob_set]
            mean = np.mean(values)
            vmin = np.min(values)
            vmax = np.max(values)
            
            x = region_count(prob_set[0])
            y = mean
            yerr = [[mean - vmin], [vmax - mean]]
            
            label = None if legend_designated else f"{flt}"
            legend_designated = True
            plt.errorbar(
                x, y, yerr=yerr,
                fmt='o', color=float_to_color[flt], markersize = 3, capsize=5,
                label=label)

    plt.xlabel("# of regions")
    plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
    plt.ylabel(f"Convergence Point {error:.2g} " + hydro)
    # plt.title("Grouped Dicts Plot")
    plt.legend(title=function_name, bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True)
    plt.show()
```

```python
def plot_regions_grid(data_dicts, m0s, color_func, plots_per_row=10, figsize_per_plot=(2, 2)):
    """
    Plot a grid of region plots for a list of data_dicts.
    Each row contains up to plots_per_row subplots.
    """
    total = len(data_dicts)
    rows = math.ceil(total / plots_per_row)
    cols = plots_per_row

    fig, axes = plt.subplots(rows, cols, figsize=(figsize_per_plot[0] * cols, figsize_per_plot[1] * rows), squeeze=False)

    for idx, data_dict in enumerate(data_dicts):
        row = idx // cols
        col = idx % cols
        ax = axes[row][col]

        a = data_dict["a"]
        d = data_dict["d"]
        h = data_dict["h"]
        m0 = m0s[idx]

        # Set limits
        ax.set_xlim(0, 2 * a[-1])
        ax.set_ylim(-h, 0)
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_facecolor('white')

        # Determine color
        body_color = color_func(data_dict, m0)

        # Draw regions
        x_edges = [0] + [a[i] for i in range(len(d))]
        for i in range(len(d)):
            if data_dict["region"] == i:
              region_color = (0, 0.7, 0.125) # Green, not in the plasma colormap
            else: region_color = body_color
            x0 = x_edges[i]
            width = x_edges[i+1] - x0
            height = d[i]
            rect = Rectangle((x0, -height), width, height, facecolor=region_color, edgecolor=None)
            ax.add_patch(rect)

        # Label with m0 value
        ax.set_title(f"m0 = {m0 : .4g}", fontsize=8)

    # Hide any unused axes
    for idx in range(total, rows * cols):
        fig.delaxes(axes.flatten()[idx])

    plt.tight_layout()
    plt.show()
```
